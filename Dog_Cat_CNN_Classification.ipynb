# Development of a Convolutional Neural Network (CNN) for Dog-Cat Image Classification Using Three Convolutional Layers

**Objective:** Build, train, evaluate, and deploy a CNN model classifying images into **Dog** and **Cat** with strictly **three convolutional layers**.

### Technical Specifications:
- **Architecture:** 3x Conv2D (32 -> 64 -> 128 filters, 3x3 kernel, ReLU) + MaxPooling2D (2x2) + Flatten + Dense (128, ReLU) + Dropout (0.5) + Dense (1, Sigmoid)
- **Loss:** Binary Cross-Entropy
- **Optimizer:** Adam
- **Metric:** Accuracy
- **Input Shape:** 128x128x3
- **Dataset Split:** 80% Train | 10% Validation | 10% Test
